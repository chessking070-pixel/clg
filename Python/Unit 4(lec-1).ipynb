# String specific Funtions

```python
s = "this is python"
print(s.capitalize())   #sentence case,1st capital
print(s.upper())        #all caps
print(s.title())        #title case , first char of every word caps
print(s.swapcase())     #inverse the case
```

```python
s = "this is python"
print(s.count('i'))
print(s.count('on'))
print(s.count('is'))
print(s.count('is',4,10)) #arg : (substring,start,end)
```

```python
s = "this is python"
print(s.find("i"))
print(s.find('pyt'))
print(s.find('z'))        #doesnt give error
print(s.find('is',4,20))
print(s.rfind('is'))
```

```python
s = "this is python"
print(s.index("i"))
print(s.index('pyt'))
print(s.index('z'))        #gives error
print(s.index('is',4,20))
print(s.rindex('is'))
```

```python
s = 'python'
print(s.isalpha())
s2 = 'python1'
print(s2.isalpha())     #combination gives False
print(s2.isalnum())     #combination gives True
```

```python
s = '123'
print(s.isdigit())
print(s.isnumeric())
```

```python
s = 'python is easy12'
print(s.islower())
s = 'Python is easy'
print(s.islower())
s= 'PYTHON IS EASY'
print(s.isupper())
```

```python
s = "  \n"
print(s.isspace())
```

```python
s = 'This is Python'
print(s.istitle())
s = 'This Is Python'
print(s.istitle())
```

```python
s = 'new1_r'
print(s.isidentifier())
```
