```python
#CEASER CIPHER
msg = input("Enter a msg : ")
key = int(input("Enter shift key: "))
encode = ' '
for i in msg:
    if ord(i)>64 and ord(i)<91:    #ord is used to convert character to ASCII(int kinda) this is for CAPS
        new = ord(i)+key
        if new>90:
            new = new-26
        encode += chr(new)
    elif ord(i)>96 and ord(i)<123:   #for small(lower char)
        new = ord(i)+key
        if new>122:
            new = new-26
        encode += chr(new)
    else:                            #for special char, NO CHANGE
        encode += i
print(encode)
```

# WAP for password validation which checks for following conditions
1. It should've atleast 1 lower case char
2. Atleast One upper case char
3. Atleast one digit
4. Atleast one special char
5. The length of the pass should be between 8-16 chars

```python
password = input("Enter your pass : ")
lower = spe = dig = upper = 0
if len(password)<17 and len(password)>7:
    for i in password:
        if i.isalnum():
            if i.islower() == True:
                lower+=1
            elif i.isupper() == True:
                upper+=1
            elif i.isdigit() == True:
                dig+=1
        elif i.isspace():
            pass
        else:
            spe += 1
    if upper!=0 and lower!=0 and dig!=0 and spe!=0:
        print("Password is Valid")
    else:
        print("Password is invalid")
else:
    print("Password is invalid")
```

# WAP to check if a string symmetrical or not, a string is called symmetrical if both halves of string are same


```python
s = input("Enter a string : ")
length = len(s)
ln = length//2
if length%2==0:
    half1 = s[:ln]
    half2 = s[ln:]
else : 
    half1 = s[:ln]
    half2 = s[ln+1:]
if half1==half2:
    print('Symmetric')
else:
    print('asymmetric')
    
```

# wap to check if two strings are balanced, strings s1 and s2 are balanced if all the chars in the s1 are present in s2 the chars position doesnt matter 

```python
s1 = input('Enter a string 1 : ')
s2 = input('Enter a string 2 : ')
flag = True
for i in s1:
    if i in s2:
        continue
    else:
        flag=False
if flag:
    print("balanced")
else:
    print("not balanced")
```
