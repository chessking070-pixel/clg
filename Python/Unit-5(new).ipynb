```python
#candy

rating = [5,2,4,7,1,3,5,8,3,6]
candy = [1]*len(rating)
for i in range(1,len(rating)):
    if rating[i]>rating[i-1]:
        candy[i] = candy[i-1]+1

for i in range(len(rating)-2,-1,-1):
    if rating[i]>rating[i+1]:
        candy[i] = max(candy[i+1]+1,candy[i])
print(sum(candy))
```

```python
days = ['Mon','Tue','Wed','Thu','Fri']
steps = [252,125,657,145,236]
d = {i:j for i,j in zip(days,steps) if j<500}
print(d)

# OR 

d1={}
for i,j in zip(days,steps):
    if j<500:
        d1[i]=j
print(d1)
```

```python
books_data = [ 
"Inferno,Dan Brown,Thriller,450,4.5", 
"Atomic Habits,James Clear,SelfHelp,550,4.8", 
"Ikigai,Hector Garcia,SelfHelp,300,4.2", 
"Dune,Frank Herbert,ScienceFiction,600,4.6", 
"1984,George Orwell,ScienceFiction,500,4.7", 
"Sapiens,Yuval Noah Harari,History,650,4.9", 
"The Hobbit,J.R.R. Tolkien,Fantasy,480,4.8", 
"Becoming,Michelle Obama,Biography,700,4.7", 
"The Alchemist,Paulo Coelho,Fiction,350,4.3", 
"To Kill a Mockingbird,Harper Lee,Fiction,400,4.6" 
]

d={}
for i in books_data:
    data = i.split(',')
    d[data[0]] = {'Author':data[1],'Genre':data[2],'Price' : int(data[3]), 'Rating' :float(data[4])}
print(d)
```

```python
total = 0
count = 0
for i in d:
    price = d.get(i)['Price']
    total += price
    count +=1
print(f"Average Price of all books: {total/count}")
```

```python
maxrating=0
for i,j in d.items():
    if j['Rating']>maxrating:
        maxrating=j['Rating']
        title = i
        author = j['Author']
print(f"Highest rated book : {title} by {author} (Rating : {maxrating})")
```

```python
g = {}
for i in d:
    if d[i]['Genre'] not in g:
        g[d[i]['Genre']] = 1
    else:
        g[d[i]['Genre']] += 1
print("Books by Genre : ")
for i,j in  g.items():
    print(i + ':' , j)
```

```python
for i in d:
    if d[i]['Genre']=="ScienceFiction":
        title = i
        rating = d[i]['Rating']
        print(f"{title} - {rating}")
```

```python
scifi = {}
for i,j in d.items():
    if j['Genre'] == 'ScienceFiction':
        scifi[i] = j['Rating']
print(scifi)
```
