# Lambda Function

```python
def add(x):
    return x+2
print(add(5))
```

```python
a = lambda x:x+2
print(a(5))
```

```python
a = lambda x,y:x*y
print(a(4,5))
```

```python
a = lambda x: 'even' if x%2==0 else 'odd'
print(a(3))
```

```python
a(90)
```

```python
a = lambda x=10,y=20,z=5: x+y-z
print(a(2,6,3))
print(a(z=2,x=6,y=3))
print(a())
```

# Map,Filter,Reduce

```python
L = [10,40,56,27,13,15,70]
L1 = []
for i in L:
    L1.append(i*2)
print(L1)
```

```python
#map : changes value
L = [10,40,56,27,13,15,70]
result = list(map(lambda x:x*2,L))
print(result)
```

```python
L = [1,4,3,2,5,6,8,5]
result = list(map(lambda x: 'even' if x%2==0 else 'odd',L))
print(result)
```

```python
#filter : can remove a value
L = [10,40,56,27,13,15,70]
result = list(filter(lambda x:x%5==0,L))
print(result)
```

```python
L = ['apple','cherry','pineapple','greenapple','guava']
result = list(filter(lambda x:x[0]=='g',L))
result1 = list(filter(lambda x:x.find('apple')!=-1,L))
print(result)
print(result1)
```

```python
L = [1,4,5,7,13,67,94]
result = list(filter(lambda x:x%2==0,map(lambda x:x*2,L)))
print(result)
```

```python
#reduce
from functools import reduce
L = [3,6,7,2,8]
result = reduce(lambda x,y:x*y,L)
print(result)
```

```python
L = [23,11,45,10,1]
result = reduce(lambda x,y:x if x>y else y,
                filter(lambda x:x%2==0,list(map(lambda x:x*2,L))))
print(result)
```

```python
L = [3,6,8,2,7,9]
result = reduce(lambda x,y:x+y,
               map(lambda x:True if x%3==0 else False,map(lambda x:x+x,L)))
print(result)
```
