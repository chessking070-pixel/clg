# WAP to accept a tuple of any integer values from user and find maximum odd value max even value minimum odd value, min even value from the given tuple

```python
t = eval(input("Enter a tuple of number : "))
sortedT = sorted(t)
length = len(sortedT)
revSor = tuple(reversed(sortedT))
print(sortedT)
print(revSor)
for i in range (0,length):
    if sortedT[i]%2==0:
        print("min even value",sortedT[i])
        break
for i in range (0,length):
    if sortedT[i]%2!=0:
        print("min odd value",sortedT[i])
        break

for i in range (0,length):
    if revSor[i]%2==0:
        print("max even value",revSor[i])
        break
for i in range (0,length):
    if revSor[i]%2!=0:
        print("max odd value",revSor[i])
        break
```

```python
#SHORT APPROACH

t = eval(input("Enter a tuple of number : "))
t_odd = ()
t_even = ()
for i in t:
    if i%2==0:
        t_even += (i,)
    else:
        t_odd += (i,)
print(min(t_even))
print(min(t_odd))
print(max(t_even))
print(max(t_odd))
```
